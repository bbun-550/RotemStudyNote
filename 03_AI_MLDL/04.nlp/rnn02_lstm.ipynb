```python
# LSTM 이해 : 다음 숫자 예측
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Input, Dense
```

```python
x = np.array([[1,2,3], [2,3,4], [3,4,5], [4,5,6], [5,6,7], [20,30,40], [30,40,50], [40,50,60]])
y = np.array([4,5,6,7,8,50,60,70])

# LSTM에 맞게 3D 입력 형태로 변환 (sample, time_steps,features)
x = x.reshape((8,3,1))
print(x.shape, " ", y.shape)

# 모델 정의
model = Sequential()
model.add(Input(shape=(3,1)))
model.add(LSTM(10, activation='relu'))    # 10개의 셀 사용
# model.add(LSTM(10, activation='relu', return_sequences=True))
# model.add(GRU(10, activation='relu'))    # 메모리/params 적게 잡아먹음(빠르게 처리), but 긴 데이터(음성 데이터) 부적합
model.add(Dense(1, activation='linear'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='mse')

from tensorflow.keras.callbacks import EarlyStopping
es = EarlyStopping(monitor='loss', patience=1, mode='auto')

model.fit(x, y, epochs=1000, batch_size=4, callbacks=[es], verbose=2)

print(f"예측값 : {model.predict(x)}")
print(f"실제값 : {y}")
```

```python
# 새로운 값으로 예측
x_input = np.array([25, 35, 47])
x_input = x_input.reshape((1,3,1))
pred = model.predict(x_input)
print(pred.flatten())
```
