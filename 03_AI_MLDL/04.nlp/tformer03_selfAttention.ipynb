```python
# Transformer : Multi-head Attention for Text Classification
#  * 기본 Transformer 구조 (논문: Attention is All You Need)
#  구성 요소 	설명
#  Encoder	입력을 벡터로 인코딩함 (문장 → 의미)
#  Decoder	인코딩된 벡터를 기반으로 출력 생성 (예: 번역된 문장)
#  예시	        영어 → 프랑스어 번역 (Input → Encoder → Decoder → Output)
#  텍스트 생성과 같은 작업은 인코더, 디코더 구조가 필요함.	요약, 번역, 질의응답, 시 생성 등

# 목적이 문장 분류 (감정 분석)이기 때문에 디코더는 필요 없다. 감정 분석은 입력 문장을 보고 "긍정/부정"과 같은 클래스 라벨을 출력하는 문제이므로, 인코더만 있으면 됨.
# 문장 분류(감정 분석, 스팸 탐지, 주제 분류 등), 개체명 인식:입력 시퀀스 → 태깅 등은 디코더 X

# 감정 분석에 적합한 구조
#   1) Embedding → TransformerEncoder → GlobalAveragePooling1D → Dense
#   2) 또는 Embedding → LSTM/CNN → Dense 구조도 가능 (이전에 실습함)

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

# 1. 데이터 로드 및 전처리 -------------
max_words = 10000   # 사전에서 사용할 최대 단어 수
max_len = 100       # 입력 시퀀스 길이

# IMDB 데이터 (정수 시퀀스 형태로 되어 있음)
(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=max_words)

# 시퀀스 길이 맞추기 (패딩)
x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=max_len)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=max_len)
print(x_train[0])      # [1415   33    6   22   12  215   28   77   52 ...

# 2. Transformer 모델 정의 -----------
# 입력 시퀀스(Embedding 또는 이전 블록 출력)를 받아 Self-Attention → FeedForward 과정을 거쳐, 잔차 연결(Residual Connection) 후 결과를 출력하는 함수 작성.  이는 Transformer 인코더 블록의 기본 구성을 따르며, 실제 논문 구조와 유사하되 실용적으로 Conv1D를 이용해 FeedForward를 구현하였다.
def transformer_encoder(inputs, head_size, num_heads, ff_dim, dropout=0):
    # ① LayerNorm → Self-Attention
    x = layers.LayerNormalization(epsilon=1e-6)(inputs)     # 1e−6=0.000001
    x = layers.MultiHeadAttention(key_dim=head_size, num_heads=num_heads, dropout=dropout)(x, x)
    res = x + inputs   # ② 잔차 연결 (Residual Add)

    # ③ LayerNorm → FeedForward (Conv1D 버전)
    x = layers.LayerNormalization(epsilon=1e-6)(res)
    # x = layers.Dense(ff_dim, activation='relu')(x) : Dense 대신 Conv1D를 사용(GPU 활용 효율이 더 높다.)
    # Conv1D(kernel_size=1)은 Dense와 거의 같은 역할을 한다.
    x = layers.Conv1D(filters=ff_dim, kernel_size=1, activation="relu")(x)  # FFN의 첫 번째 Dense 역할
    x = layers.Dropout(dropout)(x)
    x = layers.Conv1D(filters=inputs.shape[-1], kernel_size=1)(x)   # FFN의 두 번째 Dense 역할
    # Dense와 거의 동일한 동작을 하지만 구현은 1D 컨볼루션
    # kernel_size=1 → 위치 독립적으로 Dense처럼 작동 (전혀 "local window" 아님)
    # GPU에서 더 빠르게 병렬처리 가능
    # Dropout 사이에 넣기 쉬워서 실용적으로 사용됨

    return x + res  # ④ 잔차 연결 (최종 출력)

# 단계별로 설명하겠다
# 단계	  구성	                설명
#  1)  LayerNormalization      	입력값을 정규화하여 학습 안정성 증가
#  2)  MultiHeadAttention      	입력 간의 관계 파악 (자기 자신을 기준으로 중요도 계산)
#  2)  res = x + inputs        	잔차 연결 (Residual Connection), 학습이 잘 되도록 돕는 핵심 구조
#  3)  Conv1D (kernel_size=1)	Dense와 동일 역할. FeedForward를 Conv1D로 구현. 위치 독립적
#  3)  Dropout                   과적합 방지
#  4)  + res	                또 한 번의 Residual 연결로 정보 보존 및 gradient 흐름 개선

# Transformer 인코더 기반 감정 분류 모델을 구성
# 입력된 문장을 감정(긍정/부정)으로 분류하는 모델을 정의. 디코더는 X
def build_sentiment_model(input_shape, head_size, num_heads, ff_dim,
                    num_transformer_blocks, mlp_units, dropout=0, mlp_dropout=0):
    # inputs: 정수 인덱스 시퀀스(예: IMDB 리뷰)를 입력받음. (input_shape = (문장 길이,))
    inputs = keras.Input(shape=input_shape)

    # 임베딩 + 노이즈 : 단어 인덱스를 64차원 벡터로 매핑 + 간단한 위치 정보
    x = layers.Embedding(input_dim=max_words, output_dim=64)(inputs)
    # Lambda: 위치 인코딩 대용으로 노이즈를 더함 (간단한 방식)
    x = layers.Lambda(lambda t: t + tf.random.normal(shape=tf.shape(t)))(x)

    # 여러 개의 인코더 블록(transformer_encoder)을 반복하여 문맥을 정제
    for _ in range(num_transformer_blocks):
        x = transformer_encoder(x, head_size, num_heads, ff_dim, dropout)
        # transformer_encoder() 함수는 Self-Attention + FFN(FeedForward Network) 구조

    # 토큰 차원을 평균 내어 문장의 전체 의미 벡터를 생성 (→ 고정 크기 벡터)
    x = layers.GlobalAveragePooling1D()(x)

    # 여러 Fully Connected Layer를 쌓아서 분류기 구성. GlobalAverage → Dense → Dropout
    for dim in mlp_units:
        x = layers.Dense(dim, activation="relu")(x)
        x = layers.Dropout(mlp_dropout)(x)

    # 0~1 사이 확률로 출력 → 이진 감정 분류 (긍정:1 / 부정:0)
    outputs = layers.Dense(1, activation="sigmoid")(x)
    return keras.Model(inputs=inputs, outputs=outputs)

# 전체 구조 요약: 감정 분석을 위한 인코더 기반 모델
  # [Input Layer] → [Embedding Layer] → [Position Noise (Lambda)]
  #   → [Transformer Encoder Block] x N
  #   → [GlobalAveragePooling1D]
  #   → [Dense Layer(s)]
  #   → [Dropout(s)]
  #   → [Sigmoid Output]

# 3. 모델 생성 및 학습 -------------
# 이 모델은 Transformer Encoder 기반으로 문장을 받아 감정을 이진 분류(긍정 vs 부정) 하는 구조
model = build_sentiment_model(
    # 입력은 정수 시퀀스 (ex: [42, 103, 8, 67, ...]) 형태.
    input_shape=(max_len,),   # 한 문장의 최대 길이는 max_len 단어
    # 한 헤드에서 사용하는 attention 벡터의 차원. 256차 벡터로 단어 관계를 파악
    head_size=256,
    # 멀티헤드 어텐션에서 병렬로 attention을 수행하는 "Head 수". 4개의 서로 다른 관점
    num_heads=4,
    # Transformer 내부의 Feed Forward Layer의 은닉 크기. 작게 설정되어 빠름
    ff_dim=4,
    # Transformer 블록을 몇 개 쌓을 것인지. 문장의 복잡한 구조를 반복적으로 처리
    num_transformer_blocks=4,
    # Transformer 이후에 오는 MLP(Dense 층). 128차의 은닉층 하나 사용
    mlp_units=[128],
    # Transformer 내부 dropout 비율. 과적합 방지
    dropout=0.25,
    # 	MLP 부분에서의 dropout 비율. 역시 regularization 용도
    mlp_dropout=0.25
)
# 동작 흐름 요약
  # 입력: [5, 10, 73, 0, 0, 0, ...] ← 단어 인덱스 시퀀스
  # Embedding: 각 단어를 64차 벡터로 변환
  # 노이즈 추가 (Lambda Layer): 약간의 Gaussian noise 추가 (regularization)
  # Transformer Block × 4: 문장의 의미 구조 파악
  # GlobalAveragePooling: 문장 전체를 하나의 벡터로 요약
  # MLP: 128차의 은닉층 통과
  # 출력: sigmoid로 0~1 사이 확률 출력 → 감정 판단

model.compile(loss="binary_crossentropy", optimizer="adam", metrics=["accuracy"])
model.summary()

history = model.fit(x_train, y_train, validation_data=(x_test, y_test), epochs=5, batch_size=32)


# 4. 사용자 문장 예측 -------------
# 정수 인덱스 사전. IMDB 리뷰 단어 인덱스 로딩 (단어 → 정수 ID)
word_index = keras.datasets.imdb.get_word_index()
  # IMDB 리뷰 데이터셋에서 사용된 단어 → 정수 인덱스를 가져온다.
  # 예: 'good': 84 → 이후에 +3을 더해 인덱스로 사용됨.

# 사용자 문장을 정수 시퀀스로 변환하고 패딩하는 함수
def encode_text(text, word_index, max_len=100):
    words = text.lower().split()   # 소문자로 변환 후 띄어쓰기 기준으로 단어 분리

    # word_index.get(w, 2)는 단어가 사전에 없을 경우 OOV (Out Of Vocabulary) 처리
    # +3은 IMDB의 고정된 특수 토큰 자리 보정을 위한 것. (실제 단어는 3부터 시작)
    encoded = [word_index.get(w, 2) + 3 for w in words]  # 0: PAD, 1: START, 2: OOV → +3 보정
    return keras.preprocessing.sequence.pad_sequences([encoded], maxlen=max_len)

# 예측 테스트
test_sentences = [
    "this movie was fantastic and wonderful",
    "i hated this movie it was boring and terrible"
]

for sentence in test_sentences:
    # 문장을 인코딩하여 모델 입력으로 전달
    x_input = encode_text(sentence, word_index, max_len=max_len)
    # model.predict()는 [ [확률값] ] 구조로 반환됨 → [0][0]으로 스칼라 추출
    prob = model.predict(x_input)[0][0]
    label = "Positive" if prob > 0.5 else "Negative"
    print(f"Text        : {sentence}")
    print(f"Prediction  : {label}")
    print(f"Confidence  : {prob:.4f}\n")



# 시각화
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, roc_curve, auc

# 1. 기본 학습 결과 시각화 ----------
def plot_training_history(history):
    plt.figure(figsize=(12, 4))

    # Accuracy
    plt.subplot(1, 2, 1)
    plt.plot(history.history["accuracy"], label="Train Accuracy")
    plt.plot(history.history["val_accuracy"], label="Val Accuracy")
    plt.title("Model Accuracy")
    plt.xlabel("Epoch")
    plt.ylabel("Accuracy")
    plt.legend()
    plt.grid(True)

    # Loss
    plt.subplot(1, 2, 2)
    plt.plot(history.history["loss"], label="Train Loss")
    plt.plot(history.history["val_loss"], label="Val Loss")
    plt.title("Model Loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()


# 2. 훈련 또는 테스트 결과의 분류 성능을 시각화할 Confusion Matrix 그리기-----------
# 이진 분류 문제(예: 감정 분석)에서 예측 확률(y_pred_prob)을 받아
# 지정된 임계값(threshold)에 따라 0 또는 1로 변환한 뒤 실제 정답(y_true)과 비교하여
# 혼동 행렬을 시각적으로 출력
def plot_confusion_matrix(y_true, y_pred_prob, threshold=0.5):
    # y_pred_prob는 예측된 확률값 (예: 0.91, 0.03 등)
    # threshold보다 크면 1(Positive), 작으면 0(Negative)로 간주
    y_pred = (y_pred_prob > threshold).astype(int)
    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(5, 4))
    # annot=True: 각 셀에 숫자 표시, fmt="d": 정수 형태로 표시,
    # xticklabels, yticklabels: 범례를 명확히 “Positive” / “Negative”로 표시
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
          xticklabels=["Negative", "Positive"], yticklabels=["Negative", "Positive"])
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.show()

# 3. ROC Curve(모델의 이진 분류 성능을 평가하는 대표적인 방법) ----------
def plot_roc_curve(y_true, y_pred_prob):
    # FPR(위양성률), TPR(재현율), threshold 계산
    # y_true: 실제 정답값 (0 또는 1의 배열)
    # y_pred_prob: 예측 확률값 (0.0 ~ 1.0 사이 실수)
    fpr, tpr, thresholds = roc_curve(y_true, y_pred_prob)
    roc_auc = auc(fpr, tpr)

    plt.figure(figsize=(6, 5))
    plt.plot(fpr, tpr, label=f"AUC = {roc_auc:.4f}")
    plt.plot([0, 1], [0, 1], linestyle="--", color="gray")
    plt.title("ROC Curve")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend(loc="lower right")
    plt.grid(True)
    plt.show()

# 예측 확률값 얻기 (test dataset 기준)
y_pred_prob = model.predict(x_test)

# 시각화 실행
plot_training_history(history)
plot_confusion_matrix(y_test, y_pred_prob)
plot_roc_curve(y_test, y_pred_prob)
```
