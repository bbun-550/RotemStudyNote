```python
# 작성 1 : 단일 실행형 스크립트 (langchain x)
import google.generativeai as genai
from PIL import Image
import io
import os

from dotenv import load_dotenv
load_dotenv()
api_key = os.getenv("GOOGLE_API_KEY")
genai.configure(api_key=api_key)

model = genai.GenerativeModel("gemini-2.5-flash")

system_prompt = "당신은 그림을 보고 무슨 내용인지 설명을 잘하는 전문가입니다"
user_prompt = "제공된 이미지를 보고 무슨 내용인지 설명해 주세요"

image_path = "pic.jpeg"
with open(image_path, "rb") as f:
  image_data = f.read()
image = Image.open(io.BytesIO(image_data)).convert("RGB")

response = model.generate_content(
    [system_prompt, user_prompt, image],
    stream=False
)

print("\n생성된 설명은 ")
if response.text:
  print(response.text)
else:
  print("응답없음")
```

```python
# 작성 2 : 클래스 기반 스크립트
class MyMultiModal:
  def __init__(self, model, system_prompt="", user_prompt="") -> None:
    self.model = model
    self.system_prompt = system_prompt
    self.user_prompt = user_prompt

  def stream(self, image_path):
    with open(image_path, "rb") as f:
      image_data = f.read()
    image = Image.open(io.BytesIO(image_data)).convert("RGB")

    response = model.generate_content(
      [self.system_prompt, self.user_prompt, image],
      stream=False
    )
    return response

# 스트리밍 응답 출력 함수
def stream_responseFunc(response):
  for chunk in response:
    if chunk.text:
      print(chunk.text, end="", flush=True)    # flush=bool : 출력 스트림의 버퍼에 있는 데이터를 즉시 대상으로 전송

model2 = genai.GenerativeModel("gemini-2.5-flash")

system_prompt = "당신은 그림을 보고 시를 잘 만드는 베스트셀러 시인입니다"
user_prompt = "제공된 이미지를 보고 베스트셀러로 선정될 수 있는 시를 작성해 주세요"

mymulti = MyMultiModal(model2, system_prompt=system_prompt, user_prompt=user_prompt)

response = mymulti.stream("pic.jpeg")
print("\n생성된 시는 ")

stream_responseFunc(response)
```
