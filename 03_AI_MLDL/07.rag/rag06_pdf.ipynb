```python
from langchain_community.document_loaders import PyPDFLoader
pdf_filepath = 'sample.pdf'
loader = PyPDFLoader(pdf_filepath)
pages = loader.load()
print(type(loader), type(pages))
print(len(pages))

# print(pages[1])
print(pages[1].page_content)
print(pages[1].metadata)
```

```python
# LLM 모델로 문서 요약하기
from langchain_google_genai import ChatGoogleGenerativeAI
from google import genai
import os
from dotenv import load_dotenv
load_dotenv()

print("각 페이지 요약")
client = genai.Client(api_key=os.getenv("GOOGLE_API_KEY"))

for i, page in enumerate(pages, start=1):
    prompt = f"다음 PDF 페이지 내용을 2줄로 요약:\n\n{page.page_content}"

    response = client.models.generate_content (
        model="gemini-2.5-flash",
        contents=prompt
    )
    print(f"페이지 {i} 요약 ---")
    print(response.text)

print("\n전체 문서 요약:")
full_text = "\n".join(p.page_content for p in pages)
prompt = "다음 문서를 10줄로 자연스러운 서술형 문단으로 요약해줘. :\n" + full_text
response_all = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response_all.text)
```

```python
# 방법2) 형식이 없는 PDF 문서: UnstructuredPDFLoader 이용하여 PDF 파일 데이터 가져오기
!pip install unstructured[pdf]
```

```python
from langchain_community.document_loaders import UnstructuredPDFLoader
pdf_filepath = 'sample.pdf'
loader = UnstructuredPDFLoader(pdf_filepath, languages=["ko"])  # 전체 텍스트를 단일 문서 객체로 반환
pages = loader.load()
print(len(pages))
# print(pages)
print(pages[0].metadata)   # PDF 문서의 원본 출처 등 메타 데이터를 확인할 수 있다.
# page_content 속성을 통해 추출된 텍스트 내용을 확인할 수 있다.
print(pages[0].page_content[:100])   # 앞에서 부터 100 글자만 출력

print('----------')
# 각 테스트 조각을 별도로 분리 하기 (mode="elements", "pages", "single")
# element : 제목, 목록, text, 표 ... 단위
# single : pdf 전체를 하나의 doc
loader = UnstructuredPDFLoader(pdf_filepath, mode='elements', languages=["ko"])
# help(UnstructuredPDFLoader)
pages = loader.load()
print(len(pages))

print(pages[0])
print(pages[1])

e = pages[0]
print(f"Type: {type(e).__name__}")
print(f"Content:{e}")

# PDF요소(Document 객체 리스트) 중 앞의 5개 요소에 대해 타입, 텍스트 일부, 위치정보(coordinates) 출력
for e in pages[:5]:
    print( f"Type: {type(e).__name__}" )
    print( f"Text: {e.page_content[:50]}" )
    print( f"Coordinates: {e.metadata.get( 'coordinates', '없음' ) }")
    print("-" * 40)
```
