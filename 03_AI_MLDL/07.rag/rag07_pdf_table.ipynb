```python
import os
import pdfplumber
import pandas as pd
from langchain_community.document_loaders import PyMuPDFLoader
import re

pdf_path = "/content/물놀이안전명소.pdf"
loader = PyMuPDFLoader(pdf_path)
docs = loader.load()

full_text = "\n\n".join([doc.page_content.strip() for doc in docs if doc.page_content.strip()])
# print("추출된 문자열 : ", full_text)

# 한글만 추출
korean_text = re.sub(r'[^가-힣\s]', '', full_text)
print("추출된 한글 문자열 : ", korean_text)

print('한 개의 긴 문자열로 결합')
mytext = " ".join(line.strip() for line in korean_text.strip().splitlines() if line.strip())
print(mytext)
```

```python
# 표 추출 (pdfplumber 사용)
# pdfplumber를 이용해서 PDF에서 표만 추출하고, 추출된 표 데이터를 pandas.DataFrame 형태로 저장
import pdfplumber
import pandas as pd

pdf_path = "/content/물놀이안전명소.pdf"
data = []

with pdfplumber.open(pdf_path) as pdf:
    for i, page in enumerate(pdf.pages):
        tables = page.extract_tables()    # 현재 페이지에서 표(table) 목록 전체를 리스트로 추출

        for table in tables:
            if not table:
                continue

            for row in table[1:]:          # 첫 줄은 보통 헤더 → 생략
                if row and len(row) >= 4:       # 열(column)의 수가 4개 이상인 경우만 처리
                    # 각 셀 추출 : 시도 = 0번째, 시군구 = 1번째, 명칭 = 2~n-2번째, 유형 = 마지막
                    sido = row[0].strip()            # 시도
                    sigungu = row[1].strip()       # 시군구
                    category = row[-1].strip()     # 명칭 부분 (중간 열들)
                    name_parts = row[2:-1]       # 유형 (마지막 열)
                    # 명칭은 열이 여러개로 나뉘어 있는 경우를 대비 2번째부터 마지막 바로 앞까지 병합
                    # 예: ["문", "수", "계", "곡"] → "문 수 계 곡"
                    name = " ".join(part.strip() for part in name_parts if part)

                    data.append( {    # 각 행을 하나의 딕셔너리로 만들어 리스트 data에 추가
                        "시도": sido,
                        "시군구": sigungu,
                        "명칭": name,
                        "유형": category,
                        "page": i + 1
                    })

df = pd.DataFrame(data)
print(df)

# 유형별 빈도수 집계
type_counts = df["유형"].value_counts()
print("유형별 빈도수:\n")
print(type_counts)
```

```python
# csv, excel 파일로 저장
df.to_csv("안전명소.csv", index=False)
df.to_excel("안전명소.xlsx", index=False)
```
