```python
# 자연어 처리
# Token, Vector, Embedding
# 단어를 수치화

# OneHot Encoding (직접 작업)
datas = ["python", "lan", "program", "computer", "say"]
sorted_datas = sorted(datas)
print("정렬된 데이터 :", sorted_datas)

manual_labels = list(range(len(sorted_datas)))
print("인덱스 라벨 :", manual_labels)

print("레이블 인코딩 (수동)")
import numpy as np
onehot_manual = np.eye(len(manual_labels))    # eye 함수 : 주어진 크기의 단위 행렬 또는 대각 행렬 생성. 
print(onehot_manual)

print()
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
encoder_labels = encoder.fit_transform(datas)
print(encoder.classes_)
print(encoder_labels)

print("OneHotEncoder 사용")
from sklearn.preprocessing import OneHotEncoder
sorted_datas_2d = np.array(sorted_datas).reshape(-1, 1)
onehot_encoder = OneHotEncoder(sparse_output=False)    # ndarray 형태로 반환하시오
onehot_encoded = onehot_encoder.fit_transform(sorted_datas_2d)
print(onehot_encoded)

print("pandas 사용")
import pandas as pd
df = pd.DataFrame({"datas": sorted_datas})
onehot_df = pd.get_dummies(df, dtype=int)
print(onehot_df)
```
