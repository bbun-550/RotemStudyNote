```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, LSTM, GRU, Input
```

```python
# 순환 신경망(RNN), 구조 이해(네트워크)
model = Sequential()
model.add(Input(shape=(2, 10)))    # 입력 형태 정의
# 2 : 시퀀스(시점) 길이, 10: 각 시점에서 입력벡터 크기
# 의미 : (batchsize, 시퀀스길이, 입력벡터 크기)
# x[
# [0.1, 0.2, 0.3, ..., 0.10], # 시점 1
# [0.1, 0.2, 0.3, ..., 0.10], # 시점 2
# ]
# ht = tanh(WxXt + WhHt-1 + b)
# params = (input_dim + units) * units + units : (10 + 3) * 3 + 3 = 42

# model.add(SimpleRNN(units=3)) # 42

model.add(LSTM(units=3))    # 4 * (10 + 3 + 1) * 3 = 168 ; 4는 게이트 수(기본 게이트 4개)

model.summary()
```

```python
# batch_size 정의
model = Sequential()
model.add(Input(batch_size=8, shape=(2, 10)))    # 입력 텐서 = (8, 2, 10)
model.add(SimpleRNN(units=3))
model.summary()
# 출력 (8, 3)
# 출력 값 : 1, many to one / one to one
```

```python
model = Sequential()
model.add(Input(batch_size=8, shape=(2, 10)))    # 입력 텐서 = (8, 2, 10)
model.add(SimpleRNN(units=3, return_sequences=True))    # 출력이 여러개 ; many to one / many to many
model.summary()
```
