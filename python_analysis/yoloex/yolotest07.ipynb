```python
data = {
  'train': '/Users/bunny/Documents/hyundairotem_aimodel/python_analysis/yoloex/Aquarium/train/images',
  'val': '/Users/bunny/Documents/hyundairotem_aimodel/python_analysis/yoloex/Aquarium/valid/images',
  'test': '/Users/bunny/Documents/hyundairotem_aimodel/python_analysis/yoloex/Aquarium/test/images',
  'nc': 7,
  'names': ['fish', 'jellyfish', 'penguin', 'puffin', 'shark', 'starfish', 'stingray']
}
import yaml
with open('/Users/bunny/Documents/hyundairotem_aimodel/python_analysis/yoloex/data.yaml', 'w') as f:
  yaml.dump(data, f)

with open('/Users/bunny/Documents/hyundairotem_aimodel/python_analysis/yoloex/data.yaml', 'r') as f:
  aquarium_yaml = yaml.safe_load(f)
  display(aquarium_yaml)
```

```python
from ultralytics import YOLO
model = YOLO('yolov8n.pt')
print(type(model.names), len(model.names))

model.train(data='data.yaml', epochs=10, patience=5, batch=32, imgsz=416, device="mps", workers=0, amp=True)    # device="cpu"
# 416, 512, 640 크기
```
